# Використання RT-DETR-l із натренованими вагами
https://docs.ultralytics.com/models/rtdetr/

Модель RT-DETR-L (Real-Time DEtection TRansformer) була обрана для порівняльного аналізу як представник принципово іншої парадигми — візуальних трансформерів (Vision Transformers), адаптованих для детекції в реальному часі. На відміну від YOLO, ця архітектура використовує механізм самонавчання уваги (Self-Attention) для обробки глобального контексту зображення та відмовляється від традиційних евристик, таких як NMS (Non-Maximum Suppression). Вибір версії Large дозволяє максимально розкрити потенціал трансформерів у задачах розуміння складних сцен та цілісності об'єктів. Це дає змогу оцінити, чи виправдовує підвищення точності (особливо показника Recall) значні витрати обчислювальних ресурсів порівняно з CNN-підходом.

```python
from ultralytics import RTDETR
```

```python
model = RTDETR("rtdetr-l.pt")
```

```python
results = model.train(
    data='data/data.yaml',  
    epochs=50,                              
    imgsz=640,                              
    
    freeze=10,  
    
    batch=4,
    workers = 0,
    
    project='my_project',
    name='rtdetr_feature_extraction'
)
```

## 1. Якість роботи моделі (Metrics)

**RT-DETR-L** показала відмінні результати, дуже близькі до **YOLO11n**, а подекуди навіть кращі, хоча різниця мінімальна.

- **mAP50 = 0.962** — 96.2% точності на середній складності. Трохи нижче, ніж у YOLO11n (0.968), але все ще на топовому рівні.
- **mAP50-95 = 0.738** — майже ідентичний результату YOLO11n (0.737), що свідчить про однаково хорошу точність локалізації.
- **Precision = 0.953**, **Recall = 0.963** — RT-DETR має вищу повноту, ніж YOLO11n (0.914). Це означає, що трансформер пропускає менше об’єктів, хоча інколи це відбувається ціною трохи меншої впевненості.

---

## 2. Ресурсоємність та час (Trade-offs)

Це найяскравіша частина порівняння — RT-DETR забезпечує майже ідентичну точність, але значно дорожчий за ресурсами.

### Час навчання
- **RT-DETR-L: 7.58 годин**
- **YOLO11n: 0.76 годин**  
Тренування займало у **10 разів більше часу**.

### Швидкість роботи (Inference)
- **RT-DETR-L: 17.6 мс/картинка (~56 FPS)**
- **YOLO11n: 2.7 мс/картинка (~370 FPS)**  
YOLO працює **в 6.5 разів швидше**.

### Розмір моделі
- **RT-DETR-L: 66.3 МБ**
- **YOLO11n: 5.5 МБ**  
Трансформер набагато важчий і менш придатний для мобільних/edge пристроїв.

### Обчислювальна складність
- **RT-DETR-L: 103.5 GFLOPs**
- **YOLO11n: 6.5 GFLOPs**  
RT-DETR потребує **15 разів більше обчислень**.

RT-DETR демонструє хорошу стабільність, але має свої нюанси.

### Occlusion Sensitivity

```python
import numpy as np
import torch

model_path = '../my_project/rtdetr_feature_extraction4/weights/best.pt'  
img_path = 'C:/Users/oopas/saint-rusuriz/data/test/images/ezgif-frame-046_jpg.rf.f6c5f2c53572259f26ff7f76af6992f2.jpg' 
mask_size = 32   
stride = 8       

model = RTDETR(model_path)
img = cv2.imread(img_path)
img = cv2.resize(img, (640, 640)) 
h, w, _ = img.shape

results = model(img, verbose=False)
baseline_conf = 0
if len(results[0].boxes) > 0:
    baseline_conf = results[0].boxes.conf.max().item()

print(f"Базова впевненість моделі: {baseline_conf:.4f}")

heatmap_h = (h - mask_size) // stride + 1
heatmap_w = (w - mask_size) // stride + 1
heatmap = np.zeros((heatmap_h, heatmap_w))

print("Починаємо сканування оклюзією... Це може зайняти хвилину.")

for i, y in enumerate(range(0, h - mask_size + 1, stride)):
    for j, x in enumerate(range(0, w - mask_size + 1, stride)):
        
        masked_img = img.copy()
        
        masked_img[y:y+mask_size, x:x+mask_size] = 128  
        
        results = model(masked_img, verbose=False)
        
        new_conf = 0
        if len(results[0].boxes) > 0:
            new_conf = results[0].boxes.conf.max().item()
            
        sensitivity = baseline_conf - new_conf
        
        heatmap[i, j] = max(0, sensitivity)

heatmap_norm = cv2.normalize(heatmap, None, 0, 255, cv2.NORM_MINMAX)
heatmap_norm = np.uint8(heatmap_norm)

heatmap_resized = cv2.resize(heatmap_norm, (w, h))

heatmap_color = cv2.applyColorMap(heatmap_resized, cv2.COLORMAP_JET)

overlay = cv2.addWeighted(img, 0.6, heatmap_color, 0.4, 0)

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title("Occlusion Sensitivity (Red = Important)")
plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.show()
```

### Глобальна увага
- Гаряча зона покриває майже весь силует об’єкта — від голови до ніг.
- Це типова риса трансформерів: модель розглядає об’єкт цілісно, а не лише центральні ознаки.
- Прямокутна форма активації свідчить, що модель добре засвоїла межі bounding box.

### Робастність до фону
- Фон (рейки, небо, текст) залишається холодним синім.
- RT-DETR стабільно ігнорує шум, не реагує на водяні знаки чи написи.

### Розділення об’єктів
- Слабке синє світіння поруч, але теплова пляма чітко належить одному об’єкту.
- DETR-архітектури добре відокремлюють близько розташовані екземпляри завдяки object queries.

---

| Характеристика | YOLOv11n (CNN) | RT-DETR (Transformer) |
|----------------|----------------|------------------------|
| Характер уваги | Локальний, сфокусований на ключових ознаках | Глобальний, рівномірно покриває силует |
| Форма плями | Органічна, повторює форму тіла | Більш прямокутна, заповнює рамку |
| Висновок | Швидке рішення за ключовими ознаками | Глибший аналіз структури об’єкта |

---

RT-DETR демонструє відмінну стабільність і цілісне “сприйняття” об’єкта, що корисно при часткових перекриттях або слабкій видимості.  
Водночас це досягається **значно більшими ресурсними затратами**, порівняно з YOLO.

## Input Optimization / Feature Visualization

```python
model_path = '../my_project/rtdetr_feature_extraction4/weights/best.pt'
target_class_id = 1  # BBM
num_iterations = 200
lr = 5.0
img_size = 640

model = RTDETR(model_path) 
model_names = model.names
print(f"Target class: {model_names[target_class_id]}")

py_model = model.model.to('cuda')
py_model.eval()

input_tensor = torch.full((1, 3, img_size, img_size), 0.5).cuda()
input_tensor.requires_grad_(True)

optimizer = torch.optim.SGD([input_tensor], lr=lr)

print("Генеруємо 'ідеальний образ' класу...")

for i in range(num_iterations):
    optimizer.zero_grad()
    
    preds = py_model(input_tensor)
    
    if isinstance(preds, tuple):
        preds = preds[0] 
        
    output = preds 
    
    if output.shape[1] == 300: 
        class_idx = 4 + target_class_id
        
        class_activation = output[0, :, class_idx]
        
    else:
        class_activation = output[0, 4 + target_class_id, :] 

    loss = -class_activation.mean()
    
    loss.backward()
    
    if input_tensor.grad is not None:
        input_tensor.grad.data /= (input_tensor.grad.data.std() + 1e-5)

    optimizer.step()
    
    with torch.no_grad():
        input_tensor.clamp_(0, 1)
    
    if i % 50 == 0:
        print(f"Iter {i}: Loss {loss.item():.4f}")

res_img = input_tensor.detach().cpu().squeeze().permute(1, 2, 0).numpy()

plt.figure(figsize=(8, 8))
plt.title(f"Generated Input for '{model_names[target_class_id]}'\nModel: RT-DETR")
plt.imshow(res_img)
plt.axis('off')
plt.show()
```

Отримане зображення демонструє, що модель RT-DETR при класифікації BBM значною мірою покладається на текстурні ознаки високої частоти, а не лише на загальний силует об'єкта. Це підтверджує тезу про те, що глибокі нейромережі сприймають вхідні дані інакше, ніж людина, і вразливі до специфічних піксельних збурень (що є основою для Adversarial Attacks) .

## Gaussian Blur

```python
import torchvision.transforms.functional as TF

model_path = '../my_project/rtdetr_feature_extraction4/weights/best.pt'  
target_class_id = 1  # BBM
num_iterations = 400 
lr = 2.0             
img_size = 640
blur_every = 4       

model = RTDETR(model_path) 
model_names = model.names
print(f"Target class: {model_names[target_class_id]}")

py_model = model.model.to('cuda')
py_model.eval()

input_tensor = torch.full((1, 3, img_size, img_size), 0.5).cuda()
input_tensor.requires_grad_(True)

optimizer = torch.optim.SGD([input_tensor], lr=lr)

print("Генеруємо силует (з регуляризацією)...")

for i in range(num_iterations):
    optimizer.zero_grad()
    
    if i % blur_every == 0:
        with torch.no_grad():
            input_tensor[:] = TF.gaussian_blur(input_tensor, kernel_size=5, sigma=1.0)

    preds = py_model(input_tensor)
    
    if isinstance(preds, tuple):
        preds = preds[0]
    output = preds 
    
    if output.shape[1] == 300:
        class_idx = 4 + target_class_id
        activations = output[0, :, class_idx]
        loss = -activations.topk(10)[0].mean() 
    else:
        class_activation = output[0, 4 + target_class_id, :] 
        loss = -class_activation.mean()
    
    loss.backward()
    
    if input_tensor.grad is not None:
        input_tensor.grad.data /= (input_tensor.grad.data.std() + 1e-5)

    optimizer.step()
    
    with torch.no_grad():
        input_tensor.clamp_(0, 1)
    
    if i % 50 == 0:
        print(f"Iter {i}: Loss {loss.item():.4f}")

res_img = input_tensor.detach().cpu().squeeze().permute(1, 2, 0).numpy()

res_img = (res_img - res_img.min()) / (res_img.max() - res_img.min())

plt.figure(figsize=(10, 10))
plt.title(f"Generated Silhouette for '{model_names[target_class_id]}'\nModel: RT-DETR (with Gaussian Blur)")
plt.imshow(res_img)
plt.axis('off')
plt.show()
```

Навіть з розмиттям (Blur), модель RT-DETR (як і більшість глибоких мереж) є "лінивою". Вона знаходить найпростіший спосіб максимізувати впевненість класу BBM. Для неї набір яскравих пікселів певної частоти — це вже "100% танк", їй не потрібні колеса чи дуло, щоб бути впевненою.

## Feature Visualization via Activation Maximization with Regularization

```python
model_path = '../my_project/rtdetr_feature_extraction4/weights/best.pt'
target_class_id = 1  # BBM
num_iterations = 250 
lr = 0.1             
img_size = 640

def random_shift(tensor, shift_max=8):
    sx = np.random.randint(-shift_max, shift_max)
    sy = np.random.randint(-shift_max, shift_max)
    return torch.roll(tensor, shifts=(sy, sx), dims=(2, 3))

model = RTDETR(model_path) 
print(f"Target: {model.names[target_class_id]}")
py_model = model.model.to('cuda').eval()

input_tensor = torch.full((1, 3, img_size, img_size), 0.5).cuda()
input_tensor.requires_grad_(True)

optimizer = torch.optim.SGD([input_tensor], lr=lr, weight_decay=1e-4)

print("Генеруємо з Jitter та Decay...")

for i in range(num_iterations):
    optimizer.zero_grad()
    
    shifted_input = random_shift(input_tensor, shift_max=4)
    
    preds = py_model(shifted_input)
    if isinstance(preds, tuple): preds = preds[0]
    
    if preds.shape[1] == 300: # RT-DETR
        class_idx = 4 + target_class_id
        
        loss = -preds[0, :, class_idx].max() 
    else: 
        loss = -preds[0, 4 + target_class_id, :].max()

    loss.backward()
    
    if input_tensor.grad is not None:
        input_tensor.grad.data /= (input_tensor.grad.data.std() + 1e-5)

    optimizer.step()
    
    with torch.no_grad():
        input_tensor *= 0.99 
        if i % 2 == 0: 
            input_tensor[:] = TF.gaussian_blur(input_tensor, kernel_size=3, sigma=0.5)
        input_tensor.clamp_(0, 1)
    
    if i % 50 == 0:
        print(f"Iter {i}: Loss {loss.item():.4f}")

res_img = input_tensor.detach().cpu().squeeze().permute(1, 2, 0).numpy()
res_img = (res_img - res_img.min()) / (res_img.max() - res_img.min())

plt.figure(figsize=(10, 10))
plt.title(f"Optimized Feature: '{model.names[target_class_id]}'\n(with Jitter & Decay)")
plt.imshow(res_img)
plt.axis('off')
plt.show()
```

Використання регуляризації (Jitter + Decay) дозволило довести, що RT-DETR має поняття про "об'єкт" як про просторово обмежену сутність, тоді як YOLO більше покладається на наявність характерних текстур. Це пояснює, чому RT-DETR (як ми бачили раніше в метриках) має трохи вищий Recall (краще знаходить об'єкти) і кращу стійкість до оклюзій — він шукає "центр тяжіння" об'єкта.
